```python
def count_cross_edges(groupA, groupB, graph):
    setA = set(groupA)
    cross_edges_count = 0

    for user in groupB:
        for neighbor in graph[user]:
            if neighbor in setA:
                cross_edges_count += 1

    return cross_edges_count
```

```python
def find_balanced_partition_greedy(graph):
    total_users = list(graph.keys())
    N = len(total_users)

    min_size = (4 * N + 9) // 10

    shuffled_users = list(set(total_users))
    split_point = min_size

    groupA = shuffled_users[:split_point]
    groupB = shuffled_users[split_point:]

    current_cross = count_cross_edges(groupA, groupB, graph)
    improved = True

    while improved is True:
        improved = False
        best_move = None
        best_new_cross = current_cross

        if len(groupA) > min_size:
            for user in groupA:
                tempA = list(set(groupA) - {user})
                tempB = list(set(groupB) | {user})
                temp_cross = count_cross_edges(tempA, tempB, graph)

                if temp_cross < best_new_cross:
                    best_new_cross = temp_cross
                    best_move = ('A_to_B', user)

        if len(groupB) > min_size:
            for user in groupB:
                tempA = list(set(groupA) | {user})
                tempB = list(set(groupB) - {user})
                temp_cross = count_cross_edges(tempA, tempB, graph)

                if temp_cross < best_new_cross:
                    best_new_cross = temp_cross
                    best_move = ('B_to_A', user)

        if best_move is not None:
            direction, user = best_move
            if direction == 'A_to_B':
                groupA.remove(user)
                groupB.append(user)
            else:
                groupB.remove(user)
                groupA.append(user)

            current_cross = best_new_cross
            improved = True

    return (current_cross, groupA, groupB)
```

```python
def find_balanced_partition_local_search(graph, iterations):
    best_cross = float('inf')
    best_groupA = []
    best_groupB = []

    for _ in range(iterations):
        current_cross, groupA, groupB = find_balanced_partition_greedy(graph)

        if current_cross < best_cross:
            best_cross = current_cross
            best_groupA = groupA
            best_groupB = groupB

    return (best_cross, best_groupA, best_groupB)
```
