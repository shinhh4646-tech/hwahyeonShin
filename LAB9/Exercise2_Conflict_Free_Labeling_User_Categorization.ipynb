```python
def is_valid_labeling(labeling, graph):
  for u in range(len(graph)):
    for v in graph[u]:
      if u < v:
        if labeling[u] == labeling[v]:
          return False
  return True
```

```python
def assign_labels(k, graph):
  n = len(graph)
  labeling = [-1] * n

  def backtrack(node):
    if node == n:
      return True
    for color in range(k):
      labeling[node] = color
      valid = True
      for v in graph[node]:
        if labeling[v] != -1 and labeling[v] == labeling[node]:
          valid = False
          break
      if valid:
        if backtrack(node + 1):
          return True
      labeling[node] = -1
    return False
  success = backtrack(0)
  if success:
    return True, labeling
  else:
    return False, None
```

```python
def find_min_labels(graph):
  n = len(graph)
  for k in range(1, n + 1):
    success, labeling = assign_labels(k, graph)
    if success:
      return k, labeling
  return None, None
```
