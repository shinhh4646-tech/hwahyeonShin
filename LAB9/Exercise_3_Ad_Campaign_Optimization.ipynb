```python
#1
def maximize_reach(budget, costs, influences):
    n = len(costs)
    dp = [[0 for _ in range(budget + 1)] for _ in range(n + 1)]

    for i in range(1, n + 1):
        for w in range(budget + 1):
            if costs[i-1] <= w:
                dp[i][w] = max(dp[i-1][w], dp[i-1][w - costs[i-1]] + influences[i-1])
            else:
                dp[i][w] = dp[i-1][w]

    max_influence = dp[n][budget]
    selected_users_list = []
    curr_w = budget

    for i in range(n, 0, -1):
        if dp[i][curr_w] != dp[i-1][curr_w]:
            selected_users_list.append(i - 1)
            curr_w -= costs[i-1]

    return max_influence, sorted(selected_users_list)


# ---Test Case ---
if __name__ == "__main__":
    # Test Data
    budget = 50
    costs = [10, 20, 30]
    influences = [60, 100, 120]

    # Execute Function
    max_reach, selected = maximize_reach(budget, costs, influences)

    print(f"Target Budget: ${budget}")
    print(f"Maximized Influence: {max_reach} people")
    print(f"Selected Users (Indices): {selected}")

    # Verify Logic
    # (User 1: $20/100) + (User 2: $30/120) = $50/220 influence
    assert max_reach == 220, "Test Failed: Influence should be 220"
    print("Test Passed Successfully!")
```

```python
#2
def is_within_budget(selection, costs, budget):
    total_cost = 0

    for index in selection:
        total_cost += costs[index]

    return total_cost <= budget

# --- Test Case ---
if __name__ == "__main__":
    sample_costs = [10, 20, 30, 40]
    sample_budget = 50

    # Test 1: Within budget (10 + 20 = 30 <= 50)
    print(f"Test 1: {is_within_budget([0, 1], sample_costs, sample_budget)}") # Expected: True

    # Test 2: Over budget (20 + 40 = 60 > 50)
    print(f"Test 2: {is_within_budget([1, 3], sample_costs, sample_budget)}") # Expected: False
```

```python
#3

def fast_alternative_strategy(budget, costs, influences):
    n = len(costs)
    # ratio = influence / cost
    ratios = []
    for i in range(n):
        ratios.append((influences[i] / costs[i], i))

    ratios.sort(key=lambda x: x[0], reverse=True)

    total_influence = 0
    current_budget = budget
    selected_users = []

    for ratio, index in ratios:
        if costs[index] <= current_budget:
            total_influence += influences[index]
            current_budget -= costs[index]
            selected_users.append(index)

    return total_influence, sorted(selected_users)

# --- Comparison Test ---
if __name__ == "__main__":
    # Case where Greedy fails to find the optimum (Counterexample)
    budget = 50
    costs = [10, 30, 30]
    influences = [20, 50, 50]

    # 1. DP Solution (Exact)
    dp_inf, dp_users = maximize_reach(budget, costs, influences)

    # 2. Greedy Solution (Fast)
    # Ratio: User 0 (2.0), User 1 (1.66), User 2 (1.66)
    # Greedy picks User 0 first, then can't fit User 1 or 2. Total = 20.
    # DP picks User 1 and 2. Total = 100.
    greedy_inf, greedy_users = fast_alternative_strategy(budget, costs, influences)

    print(f"--- Strategy Comparison ---")
    print(f"DP Result    : {dp_inf} influence, Users {dp_users}")
    print(f"Greedy Result: {greedy_inf} influence, Users {greedy_users}")
    print(f"Optimality Gap: {dp_inf - greedy_inf} influence")
```
